```python
import numpy as np
import pandas as pd

foods=["Rice","Milk","Juice","Apple","Bread"]
days=["Day 1","Day 2","Day 3","Day 4","Day 5","Day 6","Day 7"]

calories=np.array([
    [220, 150, 120, 95, 180],
    [250, 160, 130, 100, 200],
    [230, 155, 110, 90, 175],
    [240, 165, 125, 85, 190],
    [260, 170, 135, 105, 210],
    [235, 160, 115, 95, 185],
    [245, 175, 140, 110, 195]
])
total_per_day = np.sum(calories, axis=1)
average_per_food = np.mean(calories, axis=0)

highest_day_index = np.argmax(total_per_day)
highest_day = days[highest_day_index]

df = pd.DataFrame(calories, columns=foods, index=days)

print("=== Weekly Calorie Intake Table ===")
print(df, "\n")

print("=== Total Calories Per Day ===")
print(pd.Series(total_per_day, index=days), "\n")

print("=== Average Calories Per Food Item (across 7 days) ===")
print(pd.Series(average_per_food, index=foods).round(2), "\n")

print(f"Day with Highest Total Calorie Intake: {highest_day} ({total_per_day[highest_day_index]} calories)")
```
