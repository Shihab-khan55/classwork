```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# For better visuals
sns.set(style="whitegrid")
```

```python
df = pd.read_csv("/content/housing_data.csv")
df
```

```python
df.describe()
```

```python
num_features = df.select_dtypes(include=['int64', 'float64']).shape[1]
print("Number of numerical features:", num_features)
```

```python
cat_features = df.select_dtypes(include=['object']).shape[1]
print("Number of categorical features:", cat_features)
```

```python
features = ['YearBuilt', 'TotalBsmtSF', 'GrLivArea', 'SalePrice']

plt.figure(figsize=(12,10))
for i, col in enumerate(features, 1):
    plt.subplot(2, 2, i)
    plt.hist(df[col], bins=30)
    plt.title(f'Histogram of {col}')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,10))
for i, col in enumerate(features, 1):
    plt.subplot(2, 2, i)
    sns.kdeplot(df[col], fill=True)
    plt.title(f'KDE of {col}')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,10))
for i, col in enumerate(features, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(y=df[col])
    plt.title(f'Boxplot of {col}')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='SaleCondition', data=df)
plt.title('Countplot of SaleCondition')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='Neighborhood', data=df)
plt.title('Countplot of Neighborhood')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df['GrLivArea'], df['SalePrice'])
plt.xlabel('GrLivArea')
plt.ylabel('SalePrice')
plt.title('GrLivArea vs SalePrice')
plt.show()
```
